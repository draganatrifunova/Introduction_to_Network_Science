Сега кога веќе ги истрениравме сите модели дојде време да направиме нивна споредба. При споредба на моделите не треба да се провери само кој модел има најдобра вредност под ROC кривата туку и колку точно, колку брзо и колку конзистентно секоја архитектура учи. Ние имаме:



*   12 GAT модели
*   12 GCN модели
*   12 GraphSAGE модели

Ајде да објасниме кои се разликите внатре во тие GAT, GCN и GraphSAGE типови на модели и зошто имаме 12 од секој тип.



> Од нашите GCN модели имаме 6 типа: со 2, 3 и 4 скриени слоеви, за секој модел од овој тип испробуваме да ставиме 32 и 128 неврони во скриениот слој. 3*2=6 па вкупно имаме 6 модели. Дополнително optimizer-от што го користиме за да ги ажурираме тежините на невронската мрежа експериментира 2 вредности за learning_rate 0.01 и 0.001. Па на тој начин имаме вкупно 2 * 6 = 12 модели.



> Состојбата кај GAT e малку поразлична од состојбата кај GCN и GraphSAGE. Овде не експериментираме со бројот на скриени слоеви. Сите GAT модели во овој проект имаат 2 скриени слоеви. Овде експериментираме со бројот на глави. Глава на внимание значи колку механизми користи моделот за да ја процени важноста на соседите. На пример ако имаме 2 глави на внимание, првата може да каже дека соседот А е најважен со веројатност од 0.6 од вкупната веројатност на сите соседи а, втората глава може да каже дека соседот Б е најважен со веројатност од 0.4 од вкупната веројатност.
Во нашите експерименти употребуваме модели со 1, 2 или 3 глави на внимание. Исто како кај другите два модели на граф невронска мрежа користиме различен број на неврони во скриениот слој но, овојпат не користиме 32 и 128 туку 64 и 128 за малку поголемо експериментирање. Оptimizer-от што го користиме за да ги ажурираме тежините на невронската мрежа експериментира 2 вредности за learning_rate 0.01 и 0.001 па за секој од овие 6 модели имаме и два типа на оптимизери.




> Од нашите GraphSAGE модели имаме 6 типа: со 2, 3 и 4 скриени слоеви, за секој модел од овој тип испробуваме да ставиме 32 и 128 неврони во скриениот слој. 3*2=6 па вкупно имаме 6 модели. Дополнително optimizer-от што го користиме за да ги ажурираме тежините на невронската мрежа експериментира 2 вредности за learning_rate 0.01 и 0.001. Па на тој начин имаме вкупно 2 * 6 = 12 модели.







```python
import pandas as pd

gat = pd.read_csv('gat_final_metrics.csv')
gcn = pd.read_csv('gcn_final_metrics (1).csv')
graphsage = pd.read_csv('graphsage_final_metrics.csv')

print(gat.shape, gcn.shape, graphsage.shape)
gat.head()
```

```python
gat["architecture"] = "GAT"
gcn["architecture"] = "GCN"
graphsage["architecture"] = "GraphSAGE"

all_metrics = pd.concat(
    [gcn, gat, graphsage],
    ignore_index=True
)

print("Total experiments:", len(all_metrics))

display(all_metrics)
```

```python
ranking = all_metrics.sort_values(
    by="best_val_auc",
    ascending=False
).reset_index(drop=True)

ranking.index = ranking.index + 1

display(ranking)
```

Овој код ги подредува сите модели од најдобар до најслаб и во него можеме да видиме дека дефинитивно кој и да е моделот learning_rate = 0.001 дава подобри резултати. Исто така од истражувањето донесено е заклучок дека  поверојатно е дека моделот што содржи 128 неврони во скриениот слој би дал подобри резултати но, тоа не е секогаш и точно.




> Што се однесува до GAT кај него можеме да забележиме дека неговиот успех во голема мера зависи од бројот на глави на внимание. Поверојатно е дека GAT со повеќе глави ќе даде поточни резултати. Но, за овој модел на невронски мрежи мора да напоменеме и една друга работа а, тоа е дека GAT го трениравме само со два скриени слоја додека GCN и GraphSAGE беа тренирани со 2, 3 и 4. Од друга страна пак не треба да забораваме дека бројот на неврони во скриениот слој беше 64 и 128 а, не 32 и 128 како што беше во другите два.



```python
architecture_comparison = all_metrics.groupby("architecture").agg(
    number_of_models=("model", "count"),
    average_val_auc=("best_val_auc", "mean"),
    best_val_auc=("best_val_auc", "max"),
    average_test_auc=("test_auc", "mean"),
    best_test_auc=("test_auc", "max")
).round(4)

display(architecture_comparison)
```

Од податоците во претходната келија можеме да забележиме дека GAT покажал најдобри просечни резултати а, GraphSAGE ја одржува кулминациската вредност односно максимумот. GCN има најлоши резултати.

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))

plt.bar(
    architecture_comparison.index,
    architecture_comparison["average_test_auc"]
)

plt.xlabel("GNN Architecture")
plt.ylabel("Average Test AUC")
plt.title("Average Test AUC: GAT vs GCN vs GraphSAGE")
plt.ylim(0.9, 1)
plt.grid(axis="y")

plt.show()
```

```python
gat_history = pd.read_csv('gat_history (2).csv')
gcn_history = pd.read_csv('gcn_history (1).csv')
graphsage_history = pd.read_csv('graphsage_history.csv')

print(gat_history.shape, gcn_history.shape, graphsage_history.shape)
gat_history
```

```python
gat_history["architecture"] = "GAT"
gcn_history["architecture"] = "GCN"
graphsage_history["architecture"] = "GraphSAGE"

all_history = pd.concat(
    [gcn_history, gat_history, graphsage_history],
    ignore_index=True
)

print("Total rows:", len(all_history))

display(all_history)
```

```python
best_epoch_indices = all_history.groupby(
    ["architecture", "experiment"]
)["val_auc"].idxmax()

best_epochs = all_history.loc[
    best_epoch_indices
].reset_index(drop=True)

display(
    best_epochs[
        ["architecture", "experiment", "epoch", "train_loss", "val_auc"]
    ].sort_values("val_auc", ascending=False)
)
```

Целта на претходниот код е да ја пронајде епохата во која секој од 36-те модели постигнал највисок validation AUC.

    idxmax() го враќа индексот на редот во кој вредноста е најголема.

```python
history_comparison = best_epochs.groupby("architecture").agg(
    average_best_val_auc=("val_auc", "mean"),
    std_best_val_auc=("val_auc", "std"),
    average_best_epoch=("epoch", "mean")
).round(6)

display(history_comparison)
```

Ова ќе ни покаже која архитектура во просек постигнува подобар Validation AUC кај своите 12 конфигурации и во која епоха во просек го постигнува најдобриот резултат.

```python
time_comparison = all_history.groupby("architecture").agg(
    average_train_time=("train_time_sec", "mean"),
    average_epoch_time=("epoch_total_sec", "mean"),
    total_training_time=("epoch_total_sec", "sum")
).round(3)

time_comparison["total_training_time_min"] = (
    time_comparison["total_training_time"] / 60
)

display(time_comparison)
```
